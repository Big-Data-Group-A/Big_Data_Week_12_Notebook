```python

# ==========================================
# PART 1: SQL WITH PYTHON
# ==========================================

import pandas as pd
import sqlite3

# Connect to the existing database
conn = sqlite3.connect("momo.db")

# Exercise 1.1 (a): Five largest SEND transactions above 100,000 RWF
query_1a = """
SELECT txn_id, district, amount_rwf
FROM transactions
WHERE txn_type = 'send'
AND amount_rwf > 100000
ORDER BY amount_rwf DESC
LIMIT 5;
"""

print("Exercise 1.1 (a): Top 5 send transactions")
display(pd.read_sql(query_1a, conn))


# Exercise 1.1 (b): Count flagged transactions
query_1b = """
SELECT COUNT(*) AS flagged_transactions
FROM transactions
WHERE flagged = 1;
"""

print("Exercise 1.1 (b): Flagged transactions")
display(pd.read_sql(query_1b, conn))


# Exercise 1.2 (a): Statistics per transaction type
query_1_2a = """
SELECT
    txn_type,
    COUNT(*) AS transaction_count,
    SUM(amount_rwf) AS total_amount,
    AVG(amount_rwf) AS average_amount
FROM transactions
GROUP BY txn_type
ORDER BY total_amount DESC;
"""

print("Exercise 1.2 (a): Transaction type analysis")
type_summary = pd.read_sql(query_1_2a, conn)
display(type_summary)

print("Transaction type moving the most money:",
      type_summary.iloc[0]["txn_type"])


# Exercise 1.2 (b): Total transaction amount per district
query_1_2b = """
SELECT
    district,
    SUM(amount_rwf) AS total_amount
FROM transactions
GROUP BY district
ORDER BY total_amount DESC;
"""

print("Exercise 1.2 (b): District totals")
district_summary = pd.read_sql(query_1_2b, conn)
display(district_summary)

print("Leading district:",
      district_summary.iloc[0]["district"])


# Exercise 1.3 (a): Transactions per hour using SQL
query_1_3a = """
SELECT
    hour,
    COUNT(*) AS transaction_count
FROM transactions
GROUP BY hour
ORDER BY transaction_count DESC;
"""

print("Exercise 1.3 (a): Busiest hours using SQL")
sql_hourly = pd.read_sql(query_1_3a, conn)
display(sql_hourly)

print("Busiest hour:", sql_hourly.iloc[0]["hour"])
print("Transactions:", sql_hourly.iloc[0]["transaction_count"])


# Exercise 1.3 (b): Transactions per hour using Pandas
pandas_hourly = (
    df.groupby("hour")
    .size()
    .reset_index(name="transaction_count")
    .sort_values("transaction_count", ascending=False)
    .reset_index(drop=True)
)

print("Exercise 1.3 (b): Busiest hours using Pandas")
display(pandas_hourly)

# Verify that SQL and Pandas produce the same results
sql_check = sql_hourly.sort_values("hour").reset_index(drop=True)
pandas_check = pandas_hourly.sort_values("hour").reset_index(drop=True)

print("Do SQL and Pandas agree?",
      sql_check.equals(pandas_check))

assert sql_check.equals(pandas_check), "SQL and Pandas results do not match!"

```

```python

# ==========================================
# PART 2: APIs AND JSON
# ==========================================

import json
import pandas as pd
import urllib.request

# ------------------------------------------
# Exercise 2.1: JSON BY HAND
# ------------------------------------------

# Load the provided JSON sample file
with open("week12_api_sample.json", "r", encoding="utf-8") as file:
    json_string = file.read()

# Parse JSON string into a Python dictionary
json_data = json.loads(json_string)

# Find city and readings, allowing common JSON key variations
def find_key(obj, possible_keys):
    if isinstance(obj, dict):
        for key, value in obj.items():
            if key.lower() in possible_keys:
                return value
        for value in obj.values():
            result = find_key(value, possible_keys)
            if result is not None:
                return result
    elif isinstance(obj, list):
        for item in obj:
            result = find_key(item, possible_keys)
            if result is not None:
                return result
    return None

city = find_key(json_data, {"city", "name"})
readings = find_key(json_data, {"readings"})

if isinstance(readings, list):
    # Extract numeric readings from the provided JSON
    temperatures = []

    for reading in readings:
        if isinstance(reading, (int, float)):
            temperatures.append(reading)
        elif isinstance(reading, dict):
            value = find_key(reading, {"temperature", "temp", "value"})
            if isinstance(value, (int, float)):
                temperatures.append(value)

    print("Exercise 2.1: JSON Analysis")
    print("City:", city)
    print("Number of readings:", len(readings))

    if temperatures:
        print("Average reading:", round(sum(temperatures) / len(temperatures), 2))
    else:
        print("No numeric temperature readings found.")
else:
    print("JSON structure:", json_data)
    print("Check the JSON keys against the Exercise 2.1 instructions.")


# ------------------------------------------
# Exercise 2.2: CALL A REAL WEATHER API
# ------------------------------------------

# Kigali coordinates
url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=-1.9441&longitude=30.0619"
    "&hourly=temperature_2m&forecast_days=1"
)

try:
    # Fetch live weather data
    with urllib.request.urlopen(url, timeout=15) as response:
        data = json.loads(response.read().decode("utf-8"))

    print("\nLive weather API data retrieved successfully!")

except Exception as error:
    # Use the provided sample if the API is unavailable
    print("\nLive API unavailable. Using offline sample.")
    print("Reason:", error)

    with open("week12_api_sample.json", "r", encoding="utf-8") as file:
        data = json.load(file)


# Extract time and temperature lists
# Supports standard Open-Meteo JSON and a readings-based sample
if "hourly" in data:
    times = data["hourly"]["time"]
    temperatures = data["hourly"]["temperature_2m"]

elif "time" in data and "temperature_2m" in data:
    times = data["time"]
    temperatures = data["temperature_2m"]

elif "readings" in data:
    times = []
    temperatures = []

    for reading in data["readings"]:
        if isinstance(reading, dict):
            times.append(reading.get("time", reading.get("hour")))
            temperatures.append(
                reading.get("temperature", reading.get("temp"))
            )
        else:
            temperatures.append(reading)
            times.append(None)

else:
    raise ValueError("Could not find weather time and temperature data.")


# Remove missing temperatures while keeping matching times
weather_data = [
    (time, temp)
    for time, temp in zip(times, temperatures)
    if temp is not None
]

times = [item[0] for item in weather_data]
temperatures = [float(item[1]) for item in weather_data]

# Display weather statistics
print("\nExercise 2.2: Weather Analysis")
print("Number of hours:", len(temperatures))
print("Maximum temperature:", max(temperatures))
print("Minimum temperature:", min(temperatures))

hottest_index = temperatures.index(max(temperatures))

print("Hottest hour:", times[hottest_index])


# ------------------------------------------
# Exercise 2.3: API DATA AND PANDAS
# ------------------------------------------

weather_df = pd.DataFrame({
    "time": times,
    "temp_c": temperatures
})

print("\nExercise 2.3: Weather DataFrame")
display(weather_df)

# Temperature descriptive statistics
print("Temperature description:")
display(weather_df["temp_c"].describe())

# Count hours with temperatures >= 25 degrees Celsius
hot_hours = weather_df[weather_df["temp_c"] >= 25]

print("Hours with temperature >= 25°C:", len(hot_hours))
display(hot_hours)


```

```python

# ==========================================
# PART 3: PYSPARK
# ==========================================

# Install PySpark if it is not already installed
import sys
import subprocess

try:
    import pyspark
    print("PySpark is already installed.")

except ImportError:
    print("Installing PySpark. Please wait...")
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "pyspark"
    ])

# Import Spark tools
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, count, avg

# ------------------------------------------
# Start SparkSession
# ------------------------------------------

spark = (
    SparkSession.builder
    .appName("Week12_Big_Data_Lab")
    .master("local[*]")
    .getOrCreate()
)

print("SparkSession started successfully!")


# ------------------------------------------
# Exercise 3.1: LOAD AND INSPECT
# ------------------------------------------

sdf = spark.read.csv(
    "week12_transactions.csv",
    header=True,
    inferSchema=True
)

print("\nExercise 3.1: Dataset Information")

# Display number of rows
print("Number of rows:", sdf.count())

# Display schema
print("\nDataset schema:")
sdf.printSchema()

# Display first five records
print("\nFirst five records:")
sdf.show(5)


# ------------------------------------------
# Exercise 3.2 (a): FILTER TRANSACTIONS
# ------------------------------------------

large_transactions = sdf.filter(
    col("amount_rwf") > 100000
)

print("\nExercise 3.2 (a): Transactions above 100,000 RWF")

print("Number of transactions:",
      large_transactions.count())


# ------------------------------------------
# Exercise 3.2 (b): SHOW FLAGGED TRANSACTIONS
# ------------------------------------------

flagged_transactions = sdf.filter(
    col("flagged") == 1
)

print("\nExercise 3.2 (b): Five flagged transactions")

flagged_transactions.show(5)


# ------------------------------------------
# Exercise 3.3 (a): GROUP BY TRANSACTION TYPE
# ------------------------------------------

print("\nExercise 3.3 (a): Total amount per transaction type")

spark_type_summary = (
    sdf.groupBy("txn_type")
    .agg(
        count("*").alias("transaction_count"),
        sum("amount_rwf").alias("total_amount"),
        avg("amount_rwf").alias("average_amount")
    )
    .orderBy(col("total_amount").desc())
)

spark_type_summary.show()


# ------------------------------------------
# Exercise 3.3 (b): DISTRICT TOTALS
# ------------------------------------------

print("\nExercise 3.3 (b): Total amount per district")

spark_district_summary = (
    sdf.groupBy("district")
    .agg(
        sum("amount_rwf").alias("total_amount")
    )
)

# Convert Spark DataFrame into Pandas DataFrame
spark_district_pandas = spark_district_summary.toPandas()

# Sort from highest to lowest
spark_district_pandas = spark_district_pandas.sort_values(
    "total_amount",
    ascending=False
).reset_index(drop=True)

display(spark_district_pandas)


# ------------------------------------------
# Compare Spark and SQL results
# ------------------------------------------

print("\nComparing Spark and SQL transaction totals")

spark_types_sorted = (
    spark_type_summary.toPandas()
    .sort_values("txn_type")
    .reset_index(drop=True)
)

sql_types_sorted = (
    type_summary
    .sort_values("txn_type")
    .reset_index(drop=True)
)

print(
    "Do Spark and SQL transaction totals agree?",
    spark_types_sorted["txn_type"].equals(
        sql_types_sorted["txn_type"]
    )
    and (
        spark_types_sorted["total_amount"].round(2).equals(
            sql_types_sorted["total_amount"].round(2)
        )
    )
)

# Compare district totals
spark_district_sorted = spark_district_pandas.sort_values(
    "district"
).reset_index(drop=True)

sql_district_sorted = district_summary.sort_values(
    "district"
).reset_index(drop=True)

print(
    "Do Spark and SQL district totals agree?",
    spark_district_sorted["district"].equals(
        sql_district_sorted["district"]
    )
    and (
        spark_district_sorted["total_amount"].round(2).equals(
            sql_district_sorted["total_amount"].round(2)
        )
    )
)

# Comment:
# Spark and SQL should produce the same aggregation results
# because they are analyzing the same transaction dataset.


# ------------------------------------------
# BONUS: SPARK SPEAKS SQL TOO
# ------------------------------------------

# Register Spark DataFrame as a temporary SQL view
sdf.createOrReplaceTempView("transactions")

# Run the same GROUP BY SQL query used in Part 1
spark_sql_result = spark.sql("""
SELECT
    txn_type,
    COUNT(*) AS transaction_count,
    SUM(amount_rwf) AS total_amount,
    AVG(amount_rwf) AS average_amount
FROM transactions
GROUP BY txn_type
ORDER BY total_amount DESC
""")

print("\nBONUS: SQL running inside Spark")

spark_sql_result.show()

# Explanation:
# SQL can be used with different database and processing systems.
# This makes it possible to reuse familiar queries on small databases
# such as SQLite and large distributed systems such as Spark.


```

```python

# ==========================================
# PART 3: PYSPARK
# ==========================================

# Install PySpark if it is not already installed
import sys
import subprocess

try:
    import pyspark
    print("PySpark is already installed.")
except ImportError:
    print("Installing PySpark. Please wait...")
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "pyspark"
    ])

# Import Spark tools
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, count, avg

import pandas as pd

# ------------------------------------------
# Start SparkSession
# ------------------------------------------

spark = (
    SparkSession.builder
    .appName("Week12_Big_Data_Lab")
    .master("local[*]")
    .getOrCreate()
)

print("SparkSession started successfully!")

# ------------------------------------------
# Exercise 3.1: LOAD AND INSPECT
# ------------------------------------------

sdf = spark.read.csv(
    "week12_transactions.csv",
    header=True,
    inferSchema=True
)

print("\nExercise 3.1: Dataset Information")

print("Number of rows:", sdf.count())

print("\nDataset schema:")
sdf.printSchema()

print("\nFirst five records:")
sdf.show(5)

# ------------------------------------------
# Exercise 3.2 (a): FILTER TRANSACTIONS
# ------------------------------------------

large_transactions = sdf.filter(
    col("amount_rwf") > 100000
)

print("\nExercise 3.2 (a): Transactions above 100,000 RWF")

print("Number of transactions:",
      large_transactions.count())

# ------------------------------------------
# Exercise 3.2 (b): SHOW FLAGGED TRANSACTIONS
# ------------------------------------------

flagged_transactions = sdf.filter(
    col("flagged") == 1
)

print("\nExercise 3.2 (b): Five flagged transactions")

flagged_transactions.show(5)

# ------------------------------------------
# Exercise 3.3 (a): GROUP BY TRANSACTION TYPE
# ------------------------------------------

print("\nExercise 3.3 (a): Total amount per transaction type")

spark_type_summary = (
    sdf.groupBy("txn_type")
    .agg(
        count("*").alias("transaction_count"),
        sum("amount_rwf").alias("total_amount"),
        avg("amount_rwf").alias("average_amount")
    )
    .orderBy(col("total_amount").desc())
)

spark_type_summary.show()

# ------------------------------------------
# Exercise 3.3 (b): DISTRICT TOTALS
# ------------------------------------------

print("\nExercise 3.3 (b): Total amount per district")

spark_district_summary = (
    sdf.groupBy("district")
    .agg(
        sum("amount_rwf").alias("total_amount")
    )
    .orderBy(col("total_amount").desc())
)

spark_district_pandas = spark_district_summary.toPandas()

display(spark_district_pandas)

# ------------------------------------------
# COMPARE SPARK AND SQL RESULTS
# ------------------------------------------

print("\nComparing Spark and SQL transaction totals")

# Register dataset as a temporary SQL table
sdf.createOrReplaceTempView("transactions")

# Calculate transaction totals using Spark SQL
sql_type_summary = spark.sql("""
SELECT
    txn_type,
    COUNT(*) AS transaction_count,
    SUM(amount_rwf) AS total_amount,
    AVG(amount_rwf) AS average_amount
FROM transactions
GROUP BY txn_type
""").toPandas()

# Sort both results
spark_types_sorted = (
    spark_type_summary.toPandas()
    .sort_values("txn_type")
    .reset_index(drop=True)
)

sql_types_sorted = (
    sql_type_summary
    .sort_values("txn_type")
    .reset_index(drop=True)
)

# Compare transaction totals
type_totals_match = (
    spark_types_sorted["txn_type"].equals(
        sql_types_sorted["txn_type"]
    )
    and spark_types_sorted["total_amount"].equals(
        sql_types_sorted["total_amount"]
    )
)

print(
    "Do Spark and SQL transaction totals agree?",
    type_totals_match
)

# ------------------------------------------
# COMPARE DISTRICT TOTALS
# ------------------------------------------

sql_district_summary = spark.sql("""
SELECT
    district,
    SUM(amount_rwf) AS total_amount
FROM transactions
GROUP BY district
""").toPandas()

spark_district_sorted = (
    spark_district_pandas
    .sort_values("district")
    .reset_index(drop=True)
)

sql_district_sorted = (
    sql_district_summary
    .sort_values("district")
    .reset_index(drop=True)
)

district_totals_match = (
    spark_district_sorted["district"].equals(
        sql_district_sorted["district"]
    )
    and spark_district_sorted["total_amount"].equals(
        sql_district_sorted["total_amount"]
    )
)

print(
    "Do Spark and SQL district totals agree?",
    district_totals_match
)

# ------------------------------------------
# BONUS: SPARK SPEAKS SQL TOO
# ------------------------------------------



spark_sql_result = spark.sql("""
SELECT
    txn_type,
    COUNT(*) AS transaction_count,
    SUM(amount_rwf) AS total_amount,
    AVG(amount_rwf) AS average_amount
FROM transactions
GROUP BY txn_type
ORDER BY total_amount DESC
""")

spark_sql_result.show()

# ------------------------------------------
# FINAL EXPLANATION
# ------------------------------------------



print("""
Explanation:
PySpark was used to load, inspect, filter and aggregate
the transaction dataset.

Spark and SQL were compared using the same dataset.
Spark SQL was also used to execute SQL queries directly
on the Spark DataFrame.
""")
```
